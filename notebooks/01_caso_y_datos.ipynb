# 01 · Caso NorteGrid y la serie

Helios es una mesa de análisis para un gestor independiente. El expediente de demostración es **sintético**: NorteGrid (NRGX) no es un emisor real.

La serie lleva una caída plantada de aproximadamente el 18% para que el riesgo de mercado se pueda comprobar, no para que un modelo lo adivine.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "src").exists() and (ROOT.parent / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

import matplotlib.pyplot as plt
from helios.analysis.risk import compute_stats
from helios.demo.series import synthetic_ohlc
from helios.demo.truth import CHART_PATTERN, COMPANY, FIELDS, TICKER
```

## Qué hay en la ficha

Estas cifras son la verdad terreno. El PDF, las noticias y el guion salen de las mismas constantes.

```python
print(COMPANY, TICKER)
for key, value in FIELDS.items():
    print(f"{key}: {value}")
print(CHART_PATTERN)
frame = synthetic_ohlc()
stats = compute_stats(frame["close"])
print({key: round(value, 4) for key, value in stats.items()})
plt.figure(figsize=(10, 4))
plt.plot(frame["close"], color="#E0B15A")
plt.title("NRGX · cierre sintético")
plt.show()
```

## Por qué el caso está así

El margen es holgado y, a la vez, hay un permiso retrasado y un drawdown ya realizado. La regla de mesa tiene que decir **Vigilar**: no es una historia de compra ni de quiebra.

El cliente de Helios es un gestor que no quiere sacar el audio de su cliente de la máquina. Por eso el razonamiento es local y las cifras no las redacta el modelo.
