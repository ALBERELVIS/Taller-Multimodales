# 03 · Orquestación

Un solo `run_analysis` recorre el expediente. Cada etapa deja modelo, segundos y estado. Si un modelo falta, esa etapa queda en `omitido` y la tesis de control sigue publicándose.

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "src").exists() and (ROOT.parent / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from helios.demo.case import load_demo_case
from helios.orchestration.pipeline import run_analysis

case = load_demo_case()
result = run_analysis(case, use_models=True)
for stage in result.stages:
    print(f"{stage.status:8} {stage.seconds:7.1f}s  {stage.name:28} {stage.model}  {stage.detail}")
```

```python
print("modo", result.thesis_mode)
print("respaldo publicado", f"{result.published_support_ratio:.0%}")
print("frases retiradas", len(result.rejected))
print()
print(result.thesis)
print()
print(result.briefing_script)
```

Para una pasada solo con reglas y sin Ollama ni Whisper, usa `use_models=False`. La infografía y, si hay voz en el sistema, el briefing se generan igual.
